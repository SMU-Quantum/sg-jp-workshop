# The Fine City Problem

<img src="img/fine_city.svg" alt="Fine city" width="700" />

**Scenario**: The authorities are deploying specialized surveillance cameras to catch litterbugs. However, deploying and maintaning cameras have costs.
At the same time, too many cameras can cause public backlash due to lack of privacy. 
The goal is to maximize the "deterrence value" (penalties caught) while staying within a strict budget, ensuring no neighborhood is over-surveilled (which causes resident complaints).

## Problem Formulation

\begin{align*}
\max \ & \sum_{i=1}^n p_i x_i  \\
\text{s.t.} \ & \sum_{i=1}^n c_i x_i \leq B \\
& \sum_{i\in S_k} x_i \leq M_k
\end{align*}

$x_i\in\{0, 1\}$ are the decision variables indicating whether a surveillance camera is placed at location $i$ out of $n$ possible locations ($1$: placed; $0$: not placed).

- **Objective**: Maximize the expected fine collection: $\sum_{i=1}^n p_i x_i$. 
- **Budget constraint**: $\sum_{i=1}^n c_i x_i \leq B$.
- **Public backlash**: $\sum_{i\in S_k} x_i \leq M_k$.

- $p_i$: Probability of catching an offense at location $i$.
- $c_i$: Cost of installing a camera at location $i$.
- $B$: Total budget.
- $S_k$: Local cluster of neighbourhood $k$.
- $M_k$: Limit of no. of cameras in neighbourhood $k$.

This problem is similar to the *Multi-dimensional Knapsack problem*.

# Safe Distance in Hawker Centre

<img src="img/hawker_mis_before.svg" alt="Before" width="600" />

## Problem Formulation

**Scenario**: During the widespread of pandemic, the hawkers want to practise social distance of 1-meter between the tables. They want to "seal off" some the tables but keeping the maximum number of tables at the same time. The seating plan can be represented as a graph, where the each vertex denotes a table, and each edge between two tables denotes their distance is less than 1-meter. This translates to the Maximum Independent Set (MIS) problem:

\begin{align*}
\max\ & \sum_{i=1}^n x_i \\
\text{s.t.}\ & \sum_{(i,j)\in E} x_i + x_j \leq 1 
\end{align*}

$x_i\in\{0, 1\}$ are the decision variables indicating whether a table should be sealed off ($0$: sealed off; $1$: not sealed off).

If each table has a certain number of seats that we also want to maximize, we can generate the weighted-version of MIS:

\begin{align*}
\max\ & \sum_{i=1}^n w_i x_i \\
\text{s.t.}\ & \sum_{(i,j)\in E} x_i + x_j \leq 1 
\end{align*}

$w_i$ is the number of seats for Table $i$. 

**Fun fact**: MIS can be converted to the QUBO form without adding slack variables, although it has inequality constraints.

$$
\mathcal{L}(x) = -\sum_{i=1}^n w_i x_i + \sum_{(i,j)\in E} \lambda_{ij} x_ix_j 
$$

for the minimization of $\mathcal{L}(x)$. The second term (penalty) will penalize the loss function when $x_i x_j = 1$, i.e. both $x_i$ and $x_j$ are $1$'s (MIS constraint is violated).
